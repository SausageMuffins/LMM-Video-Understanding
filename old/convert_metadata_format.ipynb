```python
import pandas as pd 



# read video_errors.csv and video_metadata.csv

video_errors = pd.read_csv('video_errors.csv')
video_metadata = pd.read_csv('video_metadata.csv')


# merge the two dataframes on the 'video_id' column
merged_df = pd.merge(video_errors, video_metadata, on='youtube_url', how='outer')

merged_df
```

```python
merged_df

# extract the id from youtube_url "https://www.youtube.com/shorts/-o7pVduTGVs"
merged_df['video_id'] = merged_df["youtube_url"].str.strip().str.split("/").str[-1].apply(lambda x: f'{x}')
merged_df['video_id'] 
# Create new video_path column where its f"videos_full/{video_id}.mp4"

merged_df['video_path'] = merged_df['video_id'].apply(lambda x: f"videos_full/{x}.mp4")
```

```python
# save as video_full_metadata.csv
merged_df.to_csv('video_full_metadata.csv', index=False, encoding='utf-8-sig')
```
